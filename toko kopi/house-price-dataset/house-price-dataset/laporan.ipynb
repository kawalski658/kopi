```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import LabelEncoder

# ==========================================
# 1. BACA DATA DENGAN BENAR (Gunakan sep=';')
# ==========================================
# Ganti 'nama_file_bank.csv' dengan nama file dataset bank kamu yang sebenarnya
df = pd.read_csv('bank.csv', sep=';')

# ==========================================
# 2. PROSES LABEL ENCODING
# ==========================================
df_lencoder = df.copy()
le = LabelEncoder()

# Mengubah semua kolom teks (object) menjadi angka agar bisa dihitung korelasinya
for col in df_lencoder.select_dtypes(include=['object']).columns:
    df_lencoder[col] = le.fit_transform(df_lencoder[col].astype(str))

# ==========================================
# 3. HITUNG DAN PLOT KORELASI TERHADAP TARGET 'y'
# ==========================================
# Hitung korelasi terhadap kolom target 'y'
target_corr = df_lencoder.corr()['y']

# Urutkan dari korelasi tertinggi ke terendah berdasarkan nilai absolutnya
target_corr_sorted = target_corr.abs().sort_values(ascending=False)

# Atur ukuran gambar
plt.figure(figsize=(14, 6))

# Buat bar plot
target_corr_sorted.plot(kind='bar', color='steelblue', edgecolor='black', alpha=0.8)

# Beri judul dan label
plt.title('Correlation with Target (y)', fontsize=14)
plt.xlabel('Variables', fontsize=12)
plt.ylabel('Correlation Coefficient', fontsize=12)

# Rotasi teks sumbu X agar mudah dibaca
plt.xticks(rotation=90, ha='right')

# Tampilkan grafik
plt.tight_layout()
plt.show()
```
