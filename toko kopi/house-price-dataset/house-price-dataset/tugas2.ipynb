```python
import pandas as pd
import numpy as np

# Membuat data dummy untuk dataset kinerja mahasiswa
np.random.seed(42)
n_samples = 1200

data = {
    'Student_ID': range(1, n_samples + 1),
    'Age': np.random.randint(18, 26, size=n_samples),
    'Gender': np.random.choice(['Male', 'Female'], size=n_samples),
    'StudyHoursPerWeek': np.random.randint(2, 40, size=n_samples),
    'AttendanceRate': np.round(np.random.uniform(50.0, 100.0, size=n_samples), 2),
    'InternetAccess': np.random.choice(['Yes', 'No'], size=n_samples, p=[0.85, 0.15]),
    'ParentalEducation': np.random.choice(['High School', 'Bachelor', 'Master', 'PhD'], size=n_samples),
    'Extracurricular': np.random.choice(['Yes', 'No'], size=n_samples),
    'SleepHours': np.random.randint(4, 10, size=n_samples),
    'PreviousGrade': np.round(np.random.uniform(50.0, 100.0, size=n_samples), 2),
}

df_student = pd.DataFrame(data)

# Sengaja menambahkan beberapa missing values (NaN) agar mirip kondisi dataset riil
df_student.loc[np.random.choice(df_student.index, 150), 'StudyHoursPerWeek'] = np.nan
df_student.loc[np.random.choice(df_student.index, 80), 'ParentalEducation'] = np.nan

# Target variabel: Nilai Ujian Akhir (Final Exam Score)
df_student['FinalExamScore'] = np.round(
    0.4 * df_student['PreviousGrade'] + 
    1.2 * df_student['StudyHoursPerWeek'].fillna(15) + 
    0.3 * df_student['AttendanceRate'] + 
    np.random.normal(0, 5, size=n_samples), 2
)

# Simpan ke CSV (opsional)
df_student.to_csv('student_performance.csv', index=False)
```

```python
# Mengisi kolom numerik yang kosong dengan median
num_cols = df_student.select_dtypes(include=['number']).columns
df_student[num_cols] = df_student[num_cols].fillna(df_student[num_cols].median())

# Mengisi kolom kategorikal yang kosong dengan modus
cat_cols = df_student.select_dtypes(include=['object', 'string']).columns
for col in cat_cols:
    df_student[col] = df_student[col].fillna(df_student[col].mode()[0])
```
